Описание: Пройдитесь по списку целых чисел. Замените числа, кратные 3, на "Fizz", кратные 5 на "Buzz", кратные обоим на "FizzBuzz". Остальные числа оставьте без изменений.

```python
def fizzBuzz(array):
  arrayLength=len(array)
  for num in range(0,arrayLength):
    if array[num] % 3 == 0 and array[num] % 5 == 0:
      array[num]="FizzBuzz"
    elif array[num] % 3 == 0:
      array[num]="Fizz"
    elif array[num] % 5 == 0:
      array[num]="Buzz"
  return array
```

```python
testArray=[2,5,11,8,9,15,856,23,1,65]
```

```python
newArray=fizzBuzz(testArray)
```

Дан список температур. Найдите индексы начала и конца самой длинной непрерывной серии дней с температурой строго выше 0. Если таких дней нет, выведите -1 -1.

```python
def findLongestStreak(array):
  best_index_start =-1
  best_index_end = -1
  cur_index_start = -1
  max_len = 0
  cur_len = 0

  arrayLength=len(array)

  for index in range(0,arrayLength):
    if array[index]>0:
      if cur_len == 0:
        cur_index_start=index
      cur_len = cur_len + 1
      if cur_len>max_len:
        max_len=cur_len
        best_index_start=cur_index_start
        best_index_end=index
    else:
      cur_len=0

  return best_index_start,best_index_end
```

Дан список целых чисел. Создайте новый список, содержащий только уникальные элементы в порядке их первого появления. Не используйте встроенные методы.

```python
def removeDuplicates(array):
  qnique = []
  for elem in array:
    found =False
    for seen_item in qnique:
      if seen_item == elem:
        found = True
        break
    if not found:
      qnique.append(elem)
  return qnique
```

```python
removeDuplicates([5,5,2,1,2,7,2,1,9,7])
```

```python
arrayToShift=[2,3,4,5,6,7]
```

```python
def shiftRight(array,k):
  k = k %len(array)
  return array[-k:]+array[:-k]
```

Элемент списка считается локальным максимумом, если он строго больше обоих своих соседей. Первый и последний элементы не рассматриваются. Выведите сами пики и их индексы в исходном списке.

```python
def finLocalMax(array):
  maximums=[]
  ids=[]
  arrayLength=len(array)-1
  for i in range(1,arrayLength):
    if(array[i]>array[i+1] and array[i]>array[i-1]):
      maximums.append(array[i])
      ids.append(i)
  return maximums,ids
```

```python
arrayTest=[1,5,2,4,1,6,2,3,9]
print(finLocalMax(arrayTest))
```

Реализуйте иерархию классов для представления сотрудников в системе учета компании.

Базовый класс Employee:

	Конструктор должен принимать и сохранять основные данные: id, name, position (должность), base_salary (базовый оклад).

	Метод calculate_salary() возвращает итоговую зарплату. В базовой реализации это просто base_salary.

Дочерние классы:

	Manager: Помимо базового оклада, получает фиксированный бонус (bonus). Переопределите метод расчета зарплаты.

	Developer: Может иметь почасовую ставку (hourly_rate) и отработанные часы (hours_worked). ИЛИ получает бонус за завершенные проекты. Выберите одну схему и переопределите метод расчета зарплаты.

	(На ваше усмотрение) Придумайте и реализуйте еще один тип сотрудника .

Требования:

	Используйте наследование.

	В конструкторах дочерних классов обязательно применяйте super().

	Продемонстрируйте работу: создайте список с разными типами сотрудников и выведите зарплату для каждого, вызвав один и тот же метод calculate_salary().

```python
class Employee:
  def __init__ (self,id,name,position,base_salary):
    self.id =id
    self.name=name
    self.position=position
    self.base_salary=base_salary

  def calculateSalary(self):
    return self.base_salary

class Manager(Employee):
  def __init__ (self,id,name,position,base_salary,bonus):
    super().__init__(id,name,position,base_salary)
    self.bonus=bonus

  def calculateSalary(self):
    return self.base_salary+self.bonus


class Developer(Employee):
  def __init__ (self,id,name,position,base_salary,hourly_rate,hours_worked):
    super().__init__(id,name,position,base_salary)
    self.hourly_rate=hourly_rate
    self.hours_worked=hours_worked

  def calculateSalary(self):
    return self.base_salary + (self.hours_worked * self.hourly_rate)


class SalesPerson(Employee):
  def __init__ (self,id,name,position,base_salary,sales_amount,commision_rate):
    super().__init__(id,name,position,base_salary)
    self.sales_amount=sales_amount
    self.commision_rate=commision_rate

  def calculateSalary(self):
    return self.base_salary + (self.sales_amount * self.commision_rate)
```

```python
employee1=Employee(1,"Ivan","test1",30000)
employee2=Manager(2,"Peter","test2",30000,15000)
employee3=Developer(3,"Yelena","test3",30000,1000,40)
employee4=SalesPerson(4,"Sssha","test4",30000,25,200)
```
